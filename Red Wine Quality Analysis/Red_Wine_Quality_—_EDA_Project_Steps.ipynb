# Step - 15 Apply SMOTE Analysis

```python
from sklearn .model_selection import train_test_split
from imblearn.over_sampling import SMOTE

x=df.drop("quality",axis=1)
y=df["quality"]

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
smote = SMOTE(random_state=42)

x_train_smote, y_train_smote = smote.fit_resample(x_train, y_train)
```

```python
print("Before SMOTE:")
print(y_train.value_counts().sort_index())

print("\nAfter SMOTE:")
print(y_train_smote.value_counts().sort_index())
```

```python
sns.countplot(x=y_train_smote)

plt.title("Class Distribution After SMOTE")
plt.xlabel("Wine Quality")
plt.ylabel("Count")
plt.show()
```

# Dasboard Creation

```python
!pip install plotly -q

import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

df = pd.read_csv('/content/wineQualityReds.csv')

# Remove unnecessary index column
df = df.loc[:, ~df.columns.str.startswith('Unnamed:')]

# Remove duplicate rows
df = df.drop_duplicates().reset_index(drop=True)

df.head()
```

```python
# Correlation matrix
corr = df.corr(numeric_only=True)

# Quality distribution
quality_counts = df['quality'].value_counts().sort_index()

# Create dashboard
fig = make_subplots(
    rows=3,
    cols=2,

    specs=[
        [{"type": "indicator"}, {"type": "indicator"}],
        [{"type": "bar"}, {"type": "box"}],
        [{"type": "scatter"}, {"type": "heatmap"}]
    ],

    subplot_titles=[
        "Total Wine Samples",
        "Average Wine Quality",
        "Wine Quality Distribution",
        "Alcohol by Quality",
        "Alcohol vs Wine Quality",
        "Correlation Heatmap"
    ],

    vertical_spacing=0.12,
    horizontal_spacing=0.08
)

# Total samples
fig.add_trace(
    go.Indicator(
        mode="number",
        value=len(df),
    ),
    row=1,
    col=1
)

# Average quality
fig.add_trace(
    go.Indicator(
        mode="number",
        value=df['quality'].mean(),
        number={"valueformat": ".2f"},
    ),
    row=1,
    col=2
)

# Wine Quality Distribution

fig.add_trace(
    go.Bar(
        x=quality_counts.index,
        y=quality_counts.values,
        name="Wine Count"
    ),
    row=2,
    col=1
)

# alcohol  by quality

for quality in sorted(df['quality'].unique()):

    fig.add_trace(
        go.Box(
            y=df[df['quality'] == quality]['alcohol'],
            name=str(quality),
            boxpoints='outliers'
        ),
        row=2,
        col=2
    )

# Alcholo vs quality

fig.add_trace(
    go.Scatter(
        x=df['alcohol'],
        y=df['quality'],
        mode='markers',
        marker=dict(
            size=7,
            opacity=0.5
        ),
        name='Wine Samples'
    ),
    row=3,
    col=1
)

# correlation heatmap

fig.add_trace(
    go.Heatmap(
        z=corr.values,
        x=corr.columns,
        y=corr.columns,
        zmin=-1,
        zmax=1,
        colorscale='RdBu',
        reversescale=True,
        text=np.round(corr.values, 2),
        texttemplate='%{text}'
    ),
    row=3,
    col=2
)

# Dasboard Formatting

fig.update_layout(
    title={
        'text': '🍷 Red Wine Quality Analysis Dashboard',
        'x': 0.5
    },

    height=1400,
    width=1400,

    template='plotly_white',

    showlegend=False
)

fig.update_xaxes(
    title_text='Wine Quality',
    row=2,
    col=1
)

fig.update_yaxes(
    title_text='Number of Wines',
    row=2,
    col=1
)

fig.update_xaxes(
    title_text='Wine Quality',
    row=2,
    col=2
)

fig.update_yaxes(
    title_text='Alcohol',
    row=2,
    col=2
)

fig.update_xaxes(
    title_text='Alcohol',
    row=3,
    col=1
)

fig.update_yaxes(
    title_text='Quality',
    row=3,
    col=1
)

fig.show()
```
